## ⭐ 7.4 Pemeriksaan Outlier

Outlier merupakan observasi yang memiliki nilai jauh berbeda dibandingkan mayoritas data lainnya.

Dalam konteks analisis pembiayaan, outlier tidak selalu menunjukkan kesalahan data. Pada beberapa kasus, outlier justru dapat merepresentasikan kondisi bisnis yang nyata seperti nilai pembiayaan yang sangat besar, pendapatan yang sangat tinggi, atau tingkat risiko yang ekstrem.

Oleh karena itu, tujuan pemeriksaan outlier pada tahap ini bukan untuk langsung menghapus data, melainkan untuk memahami karakteristik distribusi data dan mengidentifikasi nilai-nilai ekstrem yang berpotensi memengaruhi hasil analisis maupun model machine learning.

Pemeriksaan difokuskan pada variabel yang memiliki relevansi tinggi terhadap risiko pembiayaan.


````python
# =====================================================
# VARIABEL NUMERIK PRIORITAS
# =====================================================

numerical_cols = [
    'LoanOriginalAmount',
    'BorrowerAPR',
    'BorrowerRate',
    'DebtToIncomeRatio',
    'ProsperScore',
    'MonthlyLoanPayment',
    'StatedMonthlyIncome'
]

df[numerical_cols].describe().T
````

````python
# =====================================================
# VISUALISASI OUTLIER
# =====================================================

plt.figure(figsize=(16,8))

for i, col in enumerate(numerical_cols, 1):
    plt.subplot(2,4,i)
    sns.boxplot(y=df[col])
    plt.title(col)

plt.tight_layout()
plt.show()
````

````python
# =====================================================
# DETEKSI OUTLIER METODE IQR
# =====================================================

outlier_summary = []

for col in numerical_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - (1.5 * IQR)
    upper_bound = Q3 + (1.5 * IQR)

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    outlier_summary.append([
        col,
        len(outliers),
        round((len(outliers)/len(df))*100,2)
    ])

outlier_df = pd.DataFrame(
    outlier_summary,
    columns=[
        'Variable',
        'Jumlah Outlier',
        'Persentase (%)'
    ]
)

outlier_df
````

### 💡 Insight Outlier
Analisis outlier menunjukkan bahwa StatedMonthlyIncome memiliki proporsi outlier tertinggi sebesar 4,98%, diikuti LoanOriginalAmount sebesar 3,86%, MonthlyLoanPayment sebesar 3,01%, dan DebtToIncomeRatio sebesar 2,36%.

Dari perspektif risiko kredit, keberadaan outlier masih dianggap wajar karena dapat merepresentasikan debitur dengan karakteristik khusus seperti pendapatan tinggi atau kebutuhan pembiayaan yang lebih besar. Selain itu, dalam konteks pembiayaan konstruksi, nilai yang sangat tinggi dapat merepresentasikan kontraktor dengan skala bisnis yang lebih besar, nilai proyek yang lebih tinggi, maupun tingkat risiko yang berbeda dibandingkan mayoritas kontraktor lainnya.

Oleh karena itu, pada tahap ini outlier dipertahankan agar karakteristik bisnis yang sesungguhnya tetap tercermin dalam proses analisis.

### 🎯 Implikasi Bisnis

Dari perspektif Risk Analyst dan Credit Analyst, observasi dengan nilai ekstrem justru dapat menjadi indikator penting dalam proses identifikasi risiko pembiayaan. ###8.1.4 Pareto Analysis — Income Range

````python
# =====================================================
# PARETO ANALYSIS - INCOME RANGE
# =====================================================

income_pareto = (
    df['IncomeRange']
    .value_counts()
    .reset_index()
)

income_pareto.columns = [
    'IncomeRange',
    'Count'
]

income_pareto['Cumulative_%'] = (
    income_pareto['Count']
    .cumsum()
    /
    income_pareto['Count'].sum()
) * 100

# Reorder columns
income_pareto = income_pareto[
    ['Count', 'IncomeRange', 'Cumulative_%']
]

income_pareto
````

### 8.1.5 Pareto Chart

````python
# =====================================================
# PARETO CHART
# =====================================================

fig, ax1 = plt.subplots(figsize=(12,6))

ax1.bar(
    income_pareto['IncomeRange'],
    income_pareto['Count']
)

ax2 = ax1.twinx()

ax2.plot(
    income_pareto['IncomeRange'],
    income_pareto['Cumulative_%'],
    marker='o'
)

ax2.axhline(
    80,
    linestyle='--'
)

plt.title('Pareto Analysis Income Range')

plt.xticks(rotation=45)

plt.show()
````

###8.4.3 Pareto Loan Status

````python
# =====================================================
# PARETO LOAN STATUS
# =====================================================

loanstatus_pareto = (
    df['LoanStatus']
    .value_counts()
    .reset_index()
)

loanstatus_pareto.columns = [
    'LoanStatus',
    'Count'
]

loanstatus_pareto['Cumulative_%'] = (
    loanstatus_pareto['Count']
    .cumsum()
    /
    loanstatus_pareto['Count'].sum()
) * 100

loanstatus_pareto = loanstatus_pareto[
    ['Count','LoanStatus','Cumulative_%']
]

loanstatus_pareto
````

###💡 Insight Pareto Analysis Loan Status

###8.4.4 Pareto Chart Loan Status

### 8.4.5 MECE-Based Mapping Kolektibilitas Kredit

Pada tahap ini dilakukan klasifikasi risiko kredit menggunakan pendekatan MECE (*Mutually Exclusive Collectively Exhaustive*).

Pendekatan MECE digunakan untuk memastikan bahwa setiap kontrak pembiayaan hanya dapat berada pada satu kategori risiko kredit (*mutually exclusive*) dan seluruh status pembiayaan yang tersedia dalam dataset telah terpetakan ke dalam kategori yang sesuai (*collectively exhaustive*).

Untuk meningkatkan relevansi terhadap industri perbankan Indonesia, status pembiayaan pada dataset Prosper dipetakan ke dalam kerangka Kolektibilitas Kredit (Kol 1–Kol 5).

Melalui pendekatan ini, hasil analisis menjadi lebih mudah diinterpretasikan dalam konteks manajemen risiko kredit, pembiayaan konstruksi, dan pengelolaan kualitas portofolio pembiayaan.

### Kerangka Mapping Kolektibilitas

| Loan Status            | Kolektibilitas        |
| ---------------------- | --------------------- |
| Current                | Kol 1 - Lancar        |
| Completed              | Kol 1 - Lancar        |
| FinalPaymentInProgress | Kol 1 - Lancar        |
| Past Due (1-15 days)   | Kol 2 - Pra NPL       |
| Past Due (16-30 days)  | Kol 2 - Pra NPL       |
| Past Due (31-60 days)  | Kol 2 - Pra NPL       |
| Past Due (61-90 days)  | Kol 2 - Pra NPL       |
| Past Due (91-120 days) | Kol 3 - NPL           |
| Past Due (>120 days)   | Kol 4 - Diragukan     |
| Defaulted              | Kol 5 - Macet         |
| Chargedoff             | Kol 5 - Macet         |
| Cancelled              | Other                 |
| Withdrawn              | Other                 |


````python
# =====================================================
# MECE-BASED MAPPING KOLEKTIBILITAS KREDIT
# =====================================================

kol_mapping = {

    # Kol 1
    'Current':'Kol 1 - Lancar',
    'Completed':'Kol 1 - Lancar',
    'FinalPaymentInProgress':'Kol 1 - Lancar',

    # Kol 2
    'Past Due (1-15 days)':'Kol 2 - Pra NPL',
    'Past Due (16-30 days)':'Kol 2 - Pra NPL',
    'Past Due (31-60 days)':'Kol 2 - Pra NPL',
    'Past Due (61-90 days)':'Kol 2 - Pra NPL',

    # Kol 3
    'Past Due (91-120 days)':'Kol 3 - NPL',

    # Kol 4
    'Past Due (>120 days)':'Kol 4 - Diragukan',

    # Kol 5
    'Defaulted':'Kol 5 - Macet',
    'Chargedoff':'Kol 5 - Macet',

    # Other
    'Cancelled':'Other',
    'Withdrawn':'Other'
}

df['Kolektibilitas'] = (
    df['LoanStatus']
    .map(kol_mapping)
)

print("✅ Mapping Kolektibilitas Berhasil")
````

````python
# =====================================================
# CEK HASIL MAPPING
# =====================================================

df[
    ['LoanStatus', 'Kolektibilitas']
].drop_duplicates().sort_values(
    by='Kolektibilitas'
)
````

````python
# =====================================================
# CEK STATUS YANG BELUM TERMAPPING
# =====================================================

df[
    df['Kolektibilitas'].isna()
]['LoanStatus'].unique()
````

Sebelum melakukan pemodelan machine learning, perlu dilakukan identifikasi terhadap variabel yang berpotensi menyebabkan data leakage.

Data leakage terjadi ketika model memperoleh informasi yang sebenarnya belum tersedia pada saat keputusan pembiayaan dibuat.

Apabila variabel tersebut digunakan dalam proses training, performa model dapat terlihat sangat tinggi namun tidak dapat digunakan pada kondisi nyata.

````python
# =====================================================
# LOGISTIC REGRESSION BALANCED
# =====================================================

lr_balanced = LogisticRegression(
    max_iter=1000,
    random_state=42,
    class_weight='balanced'
)

lr_balanced.fit(
    X_train_scaled,
    y_train
)

y_pred_balanced = lr_balanced.predict(
    X_test_scaled
)

y_prob_balanced = lr_balanced.predict_proba(
    X_test_scaled
)[:,1]

balanced_metrics = pd.DataFrame({

    'Metric' : [
        'Accuracy',
        'Precision',
        'Recall',
        'F1 Score',
        'ROC AUC'
    ],

    'Value' : [

        accuracy_score(y_test,y_pred_balanced),

        precision_score(y_test,y_pred_balanced),

        recall_score(y_test,y_pred_balanced),

        f1_score(y_test,y_pred_balanced),

        roc_auc_score(y_test,y_prob_balanced)
    ]
})

balanced_metrics
````

````python
# =====================================================
# THRESHOLD OPTIMIZATION
# =====================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

threshold_results = []

thresholds = [
    0.50,
    0.55,
    0.60,
    0.65,
    0.70,
    0.75
]

for threshold in thresholds:

    y_pred_threshold = (
        y_prob_balanced >= threshold
    ).astype(int)

    threshold_results.append([

        threshold,

        accuracy_score(
            y_test,
            y_pred_threshold
        ),

        precision_score(
            y_test,
            y_pred_threshold
        ),

        recall_score(
            y_test,
            y_pred_threshold
        ),

        f1_score(
            y_test,
            y_pred_threshold
        )
    ])

threshold_df = pd.DataFrame(

    threshold_results,

    columns=[
        'Threshold',
        'Accuracy',
        'Precision',
        'Recall',
        'F1 Score'
    ]
)

threshold_df
````

````python
# =====================================================
# THRESHOLD COMPARISON
# =====================================================

plt.figure(figsize=(10,6))

plt.plot(
    threshold_df['Threshold'],
    threshold_df['Accuracy'],
    marker='o',
    label='Accuracy'
)

plt.plot(
    threshold_df['Threshold'],
    threshold_df['Recall'],
    marker='o',
    label='Recall'
)

plt.plot(
    threshold_df['Threshold'],
    threshold_df['F1 Score'],
    marker='o',
    label='F1 Score'
)

plt.legend()

plt.title(
    'Threshold Optimization Logistic Regression'
)

plt.xlabel('Threshold')
plt.ylabel('Score')

plt.grid(True)

plt.show()
````

### 💡 Insight Threshold Optimization

Pengujian beberapa nilai threshold menunjukkan adanya trade-off antara Accuracy dan Recall. Semakin tinggi threshold yang digunakan, Accuracy cenderung meningkat namun kemampuan model dalam mendeteksi Bad Loan (Recall) menurun.

Threshold 0,60 memberikan keseimbangan terbaik dengan Accuracy sebesar 80,93%, Recall sebesar 64,26%, dan F1 Score tertinggi sebesar 50,65%. Hasil ini menunjukkan bahwa penyesuaian threshold mampu meningkatkan kemampuan deteksi risiko kredit dibandingkan model baseline tanpa mengorbankan performa klasifikasi secara berlebihan.

### 11.4.1 Threshold Optimization

Meskipun XGBoost menghasilkan Accuracy dan ROC AUC yang tinggi, kemampuan model dalam mendeteksi Bad Loan masih relatif rendah. Oleh karena itu dilakukan threshold optimization untuk mencari titik keseimbangan terbaik antara Accuracy dan Recall.

Pendekatan ini dilakukan dengan mengubah batas probabilitas klasifikasi dari nilai default 0,50 menjadi beberapa alternatif threshold sehingga model dapat lebih sensitif terhadap debitur berisiko.

````python
# =====================================================
# XGBOOST THRESHOLD OPTIMIZATION
# =====================================================

threshold_results_xgb = []

thresholds = [
    0.50,
    0.45,
    0.40,
    0.35,
    0.30,
    0.25
]

for threshold in thresholds:

    y_pred_thr = (
        y_prob_xgb >= threshold
    ).astype(int)

    threshold_results_xgb.append([

        threshold,

        accuracy_score(
            y_test,
            y_pred_thr
        ),

        precision_score(
            y_test,
            y_pred_thr
        ),

        recall_score(
            y_test,
            y_pred_thr
        ),

        f1_score(
            y_test,
            y_pred_thr
        )
    ])

threshold_xgb_df = pd.DataFrame(

    threshold_results_xgb,

    columns=[
        'Threshold',
        'Accuracy',
        'Precision',
        'Recall',
        'F1 Score'
    ]
)

threshold_xgb_df
````

````python
# =====================================================
# THRESHOLD VISUALIZATION
# =====================================================

plt.figure(figsize=(10,6))

plt.plot(
    threshold_xgb_df['Threshold'],
    threshold_xgb_df['Accuracy'],
    marker='o',
    label='Accuracy'
)

plt.plot(
    threshold_xgb_df['Threshold'],
    threshold_xgb_df['Recall'],
    marker='o',
    label='Recall'
)

plt.plot(
    threshold_xgb_df['Threshold'],
    threshold_xgb_df['F1 Score'],
    marker='o',
    label='F1 Score'
)

plt.legend()

plt.grid(True)

plt.title(
    'Threshold Optimization XGBoost'
)

plt.show()
````

### 💡 Insight Threshold Optimization XGBoost

Penyesuaian threshold berhasil meningkatkan kemampuan XGBoost dalam mendeteksi Bad Loan secara signifikan. Pada threshold 0,25, Recall meningkat dari 28,81% menjadi 66,90% dengan Accuracy tetap berada di atas 80%, yaitu sebesar 82,49%.

Hasil ini menunjukkan bahwa sebagian besar permasalahan pada XGBoost bukan berasal dari kemampuan model dalam mempelajari pola risiko, melainkan dari penggunaan threshold default yang terlalu konservatif terhadap kelas Bad Loan.

````python
bad_loan = [
    'Chargedoff',
    'Defaulted'
]

df_dashboard['LoanRisk'] = np.where(
    df_dashboard['LoanStatus'].isin(bad_loan),
    'Bad Loan',
    'Good Loan'
)
````

````python
kol_mapping = {

    'Current':'Kol 1 - Lancar',
    'Completed':'Kol 1 - Lancar',
    'FinalPaymentInProgress':'Kol 1 - Lancar',

    'Past Due (1-15 days)':'Kol 2 - Pra NPL',
    'Past Due (16-30 days)':'Kol 2 - Pra NPL',
    'Past Due (31-60 days)':'Kol 2 - Pra NPL',
    'Past Due (61-90 days)':'Kol 2 - Pra NPL',

    'Past Due (91-120 days)':'Kol 3 - NPL',

    'Past Due (>120 days)':'Kol 4 - Diragukan',

    'Defaulted':'Kol 5 - Macet',
    'Chargedoff':'Kol 5 - Macet',

    'Cancelled':'Other'
}

df_dashboard['Kolektibilitas'] = (
    df_dashboard['LoanStatus']
    .map(kol_mapping)
)
````

````python
risk_mapping = {

    'Kol 1 - Lancar':'Low Risk',
    'Kol 2 - Pra NPL':'Medium Risk',
    'Kol 3 - NPL':'High Risk',
    'Kol 4 - Diragukan':'High Risk',
    'Kol 5 - Macet':'High Risk',
    'Other':'Other'
}

df_dashboard['RiskCategory'] = (
    df_dashboard['Kolektibilitas']
    .map(risk_mapping)
)
````
